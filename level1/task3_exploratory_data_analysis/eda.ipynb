```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

plt.savefig("charts/age_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(6,4))

df["Sex"].value_counts().plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.savefig("charts/gender_distribution.png")
plt.show()
```

```python
df["Survived"].value_counts().plot(kind="bar")
plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Passengers")
plt.savefig("charts/survival_distribution.png")
plt.show()
```

```python
df["Pclass"].value_counts().plot(kind="bar")
plt.title("Passenger Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.savefig("charts/passenger_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Fare"], bins=20)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.savefig("charts/fare_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.savefig("charts/age_vs_fare_scatter.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["Fare"].dropna())
plt.title("Box Plot of Fare")
plt.ylabel("Fare")
plt.savefig("charts/fare_boxplot.png")
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.savefig("charts/correlation_matrix.png")
plt.show()
```
